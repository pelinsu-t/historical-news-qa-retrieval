## Create the collection

```python
import json
import os

inputs = ["train.json", "validation.json", "test.json"]
output = "document_collection.json"

def load_list_or_empty(path):
    if not os.path.exists(path) or os.path.getsize(path) == 0:
        print(f"Skipping {path} because it is missing or empty")
        return []
    try:
        with open(path, "r", encoding="utf-8") as f:
            data = json.load(f)
        if isinstance(data, list):
            return data
        print(f"Skipping {path} because it is not a list at the top level")
        return []
    except json.JSONDecodeError:
        print(f"Skipping {path} because it is not valid JSON")
        return []

def project(recs):
    out = []
    for r in recs:
        out.append({
            "para_id": r.get("para_id", ""),
            "context": r.get("context", ""),
            "raw_ocr": r.get("raw_ocr", ""),
            "publication_date": r.get("publication_date", "")
        })
    return out

all_recs = []
for p in inputs:
    recs = load_list_or_empty(p)
    print(f"Loaded {len(recs)} records from {p}")
    all_recs.extend(project(recs))

# deduplicate by para_id keeping the first one seen
uniq = {}
for rec in all_recs:
    pid = rec.get("para_id", "")
    if pid and pid not in uniq:
        uniq[pid] = rec

result = list(uniq.values())

with open(output, "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

print(f"Wrote {len(result)} records to {output}")
print(json.dumps(result[:3], indent=2))
```

## create the test queries

```python
import json
import re
import unicodedata
import string

input_file = "test.json"
output_file = "test_queries.json"

# Load the data
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

def clean_question(text):
    if not isinstance(text, str):
        return ""
    text = unicodedata.normalize("NFKC", text)
    text = re.sub(rf"[{re.escape(string.punctuation)}]", " ", text)  # remove punctuation
    text = re.sub(r"\s+", " ", text)  # collapse multiple spaces
    return text.strip()

# Extract and clean
queries = [
    {
        "query_id": item.get("query_id", ""),
        "question": clean_question(item.get("question", "")),
    }
    for item in data
]

# Sort by query_id (assuming numeric)
queries = sorted(queries, key=lambda x: int(x["query_id"]) if str(x["query_id"]).isdigit() else x["query_id"])

# Keep only the first 10,000
queries = queries[:10000]

# Save new JSON
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(queries, f, ensure_ascii=False, indent=2)

print(f"Saved {len(queries)} entries to {output_file}")
print(json.dumps(queries[:3], indent=2))
```

## conversion

We convert the ".json" file into ".jsonl" format since splade works smoother with it.

```python
# --- CONFIGURATION ---
input_file = "document_collection.json"
output_file = "document_collection.jsonl"

print(f"Starting conversion: {input_file} -> {output_file}")

# Initialize the decoder
decoder = json.JSONDecoder()

with open(input_file, 'r', encoding='utf-8') as infile, \
     open(output_file, 'w', encoding='utf-8') as outfile:

    buffer = ""
    count = 0

    while True:
        # Read a chunk of data (chunk_size can be adjusted, 4MB is safe)
        chunk = infile.read(4096 * 1024)
        if not chunk:
            break

        buffer += chunk.strip() # clean whitespace/newlines

        # While we have data in the buffer, try to decode JSON objects
        while buffer:
            try:
                # Skip commas or brackets between objects (Clean up Lists)
                while buffer and buffer[0] in [',', '[', ']']:
                    buffer = buffer[1:].lstrip()

                if not buffer:
                    break

                # Attempt to parse one valid JSON object from the front of the buffer
                obj, index = decoder.raw_decode(buffer)

                # --- PROCESS THE OBJECT ---
                # Rename keys to match Pyserini requirements
                clean_obj = {
                    "id": obj.get("para_id", obj.get("id", "doc_unknown")),
                    "contents": obj.get("context", obj.get("text", ""))
                }

                # Write to JSONL (one line per object)
                outfile.write(json.dumps(clean_obj) + "\n")

                count += 1
                if count % 1000 == 0:
                    print(f"   Converted {count} documents...", end='\r')

                # Cut off the part we just read and continue
                buffer = buffer[index:].lstrip()

            except json.JSONDecodeError:
                # We reached the end of the valid data in this chunk
                # We need to read more from the file to complete this object
                break

print(f"Done! Saved {count} documents to {output_file}")
```

## Encoding

```python
# configure the output of the encoding
input_file = "document_collection.jsonl"
output_dir = "splade_output"
output_file = os.path.join(output_dir, "docs.jsonl")

# delete old files
if os.path.exists(output_dir):
    shutil.rmtree(output_dir)
os.makedirs(output_dir, exist_ok=True)

# Load data
print("Loading documents")
if 'df' not in locals():
    df = pd.read_json(input_file, lines=True)

# Initialize the model
device = 'cuda:0' if torch.cuda.is_available() else 'cpu'
encoder = SpladeDocumentEncoder('naver/splade-cocondenser-ensembledistil', device=device)

# Processing loop
records = df.to_dict('records')
batch_buffer = []

with open(output_file, 'w') as f:
    for i, row in enumerate(tqdm(records, mininterval=1.0, desc="Encoding Docs")):

        doc_id = str(row.get('id', row.get('docno', f'doc_{i}')))
        text = str(row.get('contents', row.get('text', '')))
        if not text.strip(): continue

        try:
            # Encode
            raw_vector = encoder.encode(text)

            # Unwrap list
            if isinstance(raw_vector, list):
                raw_vector = raw_vector[0] if len(raw_vector) > 0 else {}

            # Convert Numpy -> Python Float
            safe_vector = {}
            for word, weight in raw_vector.items():
                val = float(weight)
                if not (np.isnan(val) or np.isinf(val)):
                    safe_vector[word] = val

            # Write Line
            json_line = json.dumps({
                "id": doc_id,
                "contents": text,
                "vector": safe_vector
            })
            batch_buffer.append(json_line)

            # Flush to disk every 1000 docs
            if len(batch_buffer) >= 1000:
                f.write('\n'.join(batch_buffer) + '\n')
                batch_buffer = []

                # Clear GPU memory every 5000 docs
                if i % 5000 == 0: torch.cuda.empty_cache()

        except Exception as e:
            # Using 'tqdm.write' ensures error messages don't break the progress bar
            tqdm.write(f" Skipped doc {doc_id}: {e}")

    # Final Buffer Flush
    if batch_buffer:
        f.write('\n'.join(batch_buffer) + '\n')

print(f"Encoding finished. File saved to: {output_file}")
```

# Evaluation

```python
import pytrec_eval
import pandas as pd
from tqdm import tqdm
from pyserini.search.lucene import LuceneImpactSearcher
from pyserini.encode import SpladeQueryEncoder
```

```python
# Setup searcher
try:
    # Attempt to load onto GPU
    query_encoder = SpladeQueryEncoder('naver/splade-cocondenser-ensembledistil', device='cuda:0')
    searcher = LuceneImpactSearcher("indices/splade", query_encoder=query_encoder)
except:
    searcher = LuceneImpactSearcher("indices/splade", query_encoder='naver/splade-cocondenser-ensembledistil')

# pytrec_eval requires specific dictionary formats, not DataFrames.
# Format: {qid: {docid: relevance, ...}}
qrels_dict = {}
# Iterate safely
for _, row in qrels_df.iterrows():
    qid = str(row['qid'])
    docid = str(row['docno']) # or 'docid'
    rel = int(row['label'])   # or 'rel'

    if qid not in qrels_dict:
        qrels_dict[qid] = {}
    qrels_dict[qid][docid] = rel

# Format: {qid: query_text}
# We filter to only include queries that actually exist in the Qrels (to save time)
valid_qids = set(qrels_dict.keys())
queries_dict = {}

for _, row in queries_df.iterrows():
    qid = str(row['qid']) # or 'id'
    text = str(row['query']) # or 'query'

    if qid in valid_qids:
        queries_dict[qid] = text

print(f"   - Loaded {len(qrels_dict)} valid qrels.")
print(f"   - Loaded {len(queries_dict)} queries to search.")

# run batch search
run_dict = {} # This will store results: {qid: {docid: score}}
batch_size = 128
qids_list = list(queries_dict.keys())
texts_list = list(queries_dict.values())

for i in tqdm(range(0, len(qids_list), batch_size), desc="Searching"):
    # Slice batch
    batch_ids = qids_list[i : i+batch_size]
    batch_texts = texts_list[i : i+batch_size]

    # Search
    hits = searcher.batch_search(batch_texts, batch_ids, k=1000, threads=12)

    # Convert Pyserini hits to pytrec_eval format immediately
    for qid, results in hits.items():
        run_dict[qid] = {}
        for item in results:
            run_dict[qid][item.docid] = float(item.score)


# Define the metrics
metrics = {'P_1', 'P_5', 'P_10', 'recall_5', 'recall_10', 'ndcg_cut_5', 'ndcg_cut_10', 'map'}

evaluator = pytrec_eval.RelevanceEvaluator(qrels_dict, metrics)
results = evaluator.evaluate(run_dict)


# The evaluator returns a dict of {qid: {metric: score}}. We average them.
aggregated = {m: 0.0 for m in metrics}
count = 0

for qid in results:
    count += 1
    for m in metrics:
        aggregated[m] += results[qid][m]

print("\n" + "="*30)
print(f"   RESULTS ({count} queries)")
print("="*30)

# Map pytrec_eval names to your preferred names for cleaner output
display_names = {
    'P_1': 'P_1', 'P_5': 'P_5', 'P_10': 'P_10',
    'recall_5': 'recall_5', 'recall_10': 'recall_10',
    'ndcg_cut_5': 'ndcg_cut_5', 'ndcg_cut_10': 'ndcg_cut_10', 'map': 'map'
}

# pytrec_eval uses dots (P.1), you asked for underscores (P_1).
# We print nicely.
sorted_metrics = sorted(list(metrics))
for m in sorted_metrics:
    avg_score = aggregated[m] / count if count > 0 else 0
    clean_name = m.replace('.', '_') # Convert P.5 -> P_5
    print(f"{clean_name:<15}: {avg_score:.4f}")
print("="*30)
```
